```python
from functools import partial
from pathlib import Path
from torch import tensor, long, zeros
from torch.optim import AdamW

from karpathy_series.makemore.encoding.character import CharacterSet, CharacterEncoder
from karpathy_series.makemore.data import read_character_data
from karpathy_series.gpt_from_scratch.data import SequenceDataSplit
from karpathy_series.gpt_from_scratch.batch_generation import SequentialBlockBatchGenerator
from karpathy_series.gpt_from_scratch.training import Trainer
from karpathy_series.gpt_from_scratch.components.models.attention import AttentionModel
```

```python
data_path = Path("../../data/shakespeare.txt")
text_data = read_character_data(data_path)
char_set = CharacterSet.from_text(text_data)
char_encoder = CharacterEncoder.from_charset(char_set)
data = tensor(char_encoder.encodes(text_data), dtype=long)
splitter = SequenceDataSplit(90, 10)
data_split = splitter(data)
print(f"Split: {len(data_split.training)} training, {len(data_split.validation)} validation")
```

```python
window_length = 8
embedding_dims = 32
head_dims = 8
head_count = 4
model = AttentionModel(char_encoder.size, window_length, embedding_dims, 4, head_count, head_dims)
batcher = partial(SequentialBlockBatchGenerator, block_size=window_length, batch_size=8)
optimizer = AdamW(model.parameters(), lr=1e-3)
trainer = Trainer(model, optimizer, batcher, data_split)
```

```python
trainer.train(40000, 10)
```

```python
for row in model.generate_from_init(1000):
    print(char_encoder.decode_str(row.tolist()))
```
