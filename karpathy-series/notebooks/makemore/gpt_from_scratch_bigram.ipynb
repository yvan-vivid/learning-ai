```python
from torch import tensor, long, manual_seed, randint, stack, concat, ones, rand, multinomial, zeros
from torch.nn import Module, Embedding
from torch.nn.functional import cross_entropy, softmax
from torch.optim import AdamW
from karpathy_series.makemore.util import sample_index_logits

from karpathy_series.gpt_from_scratch.data import SequenceDataSplit
from karpathy_series.gpt_from_scratch.batch_generation import SequentialBlockBatchGenerator
from karpathy_series.gpt_from_scratch.training import Trainer
from karpathy_series.gpt_from_scratch.components.models.bigram import BigramModel
```

## Data

```python
from pathlib import Path
from karpathy_series.makemore.data import read_character_data, read_blocked_character_data
```

```python
data_path = Path("../../data/shakespeare.txt")
text_data = read_character_data(data_path)
```

```python
from karpathy_series.makemore.encoding.character import CharacterSet, CharacterEncoder
```

```python
char_set = CharacterSet.from_text(text_data)
char_encoder = CharacterEncoder.from_charset(char_set)
print(f"Token set size = {char_encoder.size}")
```

```python
data = tensor(char_encoder.encodes(text_data), dtype=long)
print(f"Full data length = {len(data)}")
```

```python
splitter = SequenceDataSplit(90, 10)
data_split = splitter(data)
print(f"Split: {len(data_split.training)} training, {len(data_split.validation)} validation") 
```

```python
def render_generated(generated):
    for row in generated:
        yield "".join(char_encoder.decodes(row.tolist()))
```

```python
model = BigramModel(char_encoder.size)
batcher = lambda data: SequentialBlockBatchGenerator(data, 8, 4)
optimizer = AdamW(model.parameters(), lr=1e-3)
trainer = Trainer(model, optimizer, batcher, data_split)
```

```python
trainer.train(40000, 10)
```

```python
start = zeros(1, 1, dtype=long)
generated = model.generate(start, 200)

for out in render_generated(generated):
    print(out)
```
